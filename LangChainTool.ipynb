```python
from langchain_community.tools import DuckDuckGoSearchRun, DuckDuckGoSearchResults ,ShellTool
from langchain_community.tools import WikipediaQueryRun,GmailSendMessage,SlackSendMessage
from langchain_community.utilities import WikipediaAPIWrapper
from dotenv import load_dotenv
load_dotenv()
```

```python
from ddgs import DDGS

question = input("What do you want to search for? ")
def search_web(query):
    with DDGS() as ddgs:
        results = ddgs.text(query, max_results=5)
        for r in results:
            print(f"Title: {r['title']}\nLink: {r['href']}\nSnippet: {r['body']}\n")
    return results
search_web(question)
```

```python
search_tool = DuckDuckGoSearchRun()

ans = search_tool.invoke("news of ipl")
print(ans)


search_tool1 = DuckDuckGoSearchResults(output_format="list")
ans = search_tool1.invoke("Today dates")
for r in ans:
    print(f"Title: {r['title']}\nLink: {r['link']}\nbody {r['snippet']}")
```

```python
shell = ShellTool()

ans = shell.run("pip list")
print(ans)
```

```python
api_wrapper = WikipediaAPIWrapper(
    top_k_results=1, 
    doc_content_chars_max=1000
)

tool = WikipediaQueryRun(api_wrapper=api_wrapper)

result = tool.run("python")
print(result)
```

```python
from langchain_community.utilities import SQLDatabase

db = SQLDatabase.from_uri(
    "mysql+pymysql://root:1234@localhost:3306/todo_app"
)

result = db.run("SELECT * FROM files LIMIT 5;")
print(result)
```

```python
from langchain_groq import ChatGroq
from langchain_community.agent_toolkits import SQLDatabaseToolkit
from langchain_community.agent_toolkits import create_sql_agent

llm = ChatGroq(temperature=0,model="llama-3.1-8b-instant",)

toolkit = SQLDatabaseToolkit(db=db, llm=llm)

agent = create_sql_agent(
    llm=llm,
    toolkit=toolkit,
    verbose=True
)

agent.run("Show top 3 users")
```
